# Steady State tandem JV simulations

```python
import os, sys
try:
    import pySIMsalabim as sim
except ImportError: # add parent directory to sys.path if pySIMsalabim is not installed
    sys.path.append('..')
    import pySIMsalabim as sim
from pySIMsalabim.experiments.tandem import *

# Use the default input files shipped with the pySIMsalabim tests (pySIMsalabim/tests/sim_inputs/default):
# check that they match the SIMsalabim version and copy them into SIMsalabim/SimSS and SIMsalabim/ZimT
sys.path.append(os.path.join('..', 'pySIMsalabim', 'tests'))
from sim_inputs_setup import prepare_test_inputs, cleanup_test_inputs
prepare_test_inputs(convert=False)
```

```python
# Set the path to the simulation setup file
cwd = os.path.abspath('..')
session_path = os.path.join(cwd, 'SIMsalabim','SimSS')
simss_device_parameters = os.path.join(session_path, 'simulation_setup_tandem_test.txt')

# Set the tandem simulation parameters
lid_sub1_end = 3
lid_sub2_start = 4
rec_contact_type = 1

ret, mess = run_tandem_simu(simss_device_parameters, session_path, lid_sub1_end, lid_sub2_start, rec_contact_type)
```

```python
# Clean up the output files (comment out if you want to keep the output files)
sim.clean_all_output(session_path)
sim.delete_folders('tmp',session_path)

# If present, remove files that contain 'sub1', or 'sub2'
for file in os.listdir(session_path):
    if 'sub1' in file or 'sub2' in file:
        if os.path.isfile(os.path.join(session_path, file)):
            os.remove(os.path.join(session_path, file))

# Remove the test input files copied into the SIMsalabim folder
cleanup_test_inputs()
```
